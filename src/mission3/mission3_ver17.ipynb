## 이번에 비교하는 두 후보

- `baseline_ce`: 추가 언어 학습 없이 ver15 방식으로 증상을 학습.
- `tapt_ce`: Training의 Fit 대화 텍스트로 단어 복원 학습을 먼저 한 뒤 같은 증상 분류를 학습.

추가 언어 학습은 800회 optimizer update를 목표로 합니다. 첫 10회 속도를 측정해 완료 가능 여부를
판단하며, 그 10회도 800회에 포함합니다. Fit 적응 학습은 최대 15분, 전체 Training 재학습 단계의
적응 학습은 최대 20분이고 두 시간 모두 180분 전체 예산에 포함합니다.
정해진 추가 학습을 완료하지 못한 모델을 완료된 TAPT 모델로 사용하지 않습니다.
시간이 부족하면 TAPT 후보를 생략하거나, 완료된 후보만 사용한 사실을 결과에 표시합니다.

후보 비교 단계의 적응 학습은 **Fit 대화만** 사용합니다. Holdout과 공식 Validation의 텍스트도
적응 학습에 넣지 않습니다. TAPT가 선택되면 공개 초기 모델에서 전체 Training 텍스트 적응 학습을
새로 하고, 선택된 epoch만큼 전체 Training 정답으로 재학습합니다.
단어 복원 학습은 고정 횟수로 표본을 뽑으므로 모든 통화를 반드시 한 번씩 읽는 것은 아닙니다.
반면 완료된 전체 지도학습은 모든 Training 샘플을 사용합니다.

TAPT는 구어체·전사 표현에 적응하는 가설입니다. 텍스트에 없는 정보나 주석 불일치를 해결한다고
가정하지 않습니다. 근거: [TAPT 논문](https://aclanthology.org/2020.acl-main.740/)

두 후보의 양성 학습 가중치는 ver15의 `recall_moderate`와 같습니다.
`[고열, 구토, 두통, 복통, 어지러움, 열상, 오심, 전신쇠약, 호흡곤란]` 순서로
`[1.4, 1.8, 2.0, 1.2, 1.3, 1.0, 3.0, 1.5, 1.3]`을 사용합니다.
후보 학습은 최대 3epoch이며, 학습률 감소 길이는 기존과 같은 4epoch 기준을 유지합니다.
Training 내부 Macro F1으로 후보와 epoch를 선택하고, 시간이 허용되면 전체 Training으로 새로 학습합니다.
공식 Validation은 최종 모델을 정한 뒤 한 번 평가합니다. 개별 증상의 정밀도·재현율·F1도 저장합니다.
