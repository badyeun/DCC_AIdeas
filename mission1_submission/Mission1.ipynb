```python
from pathlib import Path

ROOT = Path("/content/drive/MyDrive/data(inho)")

parts = sorted([
    p for p in ROOT.iterdir()
    if p.is_dir()
])

print("part 개수:", len(parts))

for p in parts:
    print(p.name)
```

```python
train_wavs = []
train_labels = []

val_wavs = []
val_labels = []

for part in parts:
    base = part / "대학부 데이터"

    # Training
    train_audio_dir = base / "Training" / "1.원천데이터" / "TS_서울_구급"
    train_label_dir = base / "Training" / "2.라벨링데이터" / "TL_서울_구급"

    # Validation
    val_audio_dir = base / "Validation" / "1.원천데이터" / "VS_서울_구급"
    val_label_dir = base / "Validation" / "2.라벨링데이터" / "VL_서울_구급"

    train_wavs.extend(train_audio_dir.rglob("*.wav"))
    train_labels.extend(train_label_dir.rglob("*.json"))

    val_wavs.extend(val_audio_dir.rglob("*.wav"))
    val_labels.extend(val_label_dir.rglob("*.json"))

print("Train wav :", len(train_wavs))
print("Train json:", len(train_labels))
print("Val wav   :", len(val_wavs))
print("Val json  :", len(val_labels))
```

```python
%cd /content/drive/MyDrive/mission1

import joblib
import numpy as np
import torch

from pathlib import Path
from sklearn.metrics import accuracy_score, classification_report, f1_score
from sklearn.preprocessing import StandardScaler
from torch import nn
from torch.utils.data import DataLoader, TensorDataset


# =========================
# 설정
# =========================

CACHE_DIR = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache"
)

TRAIN_CACHE = CACHE_DIR / (
    "train_microsoft_wavlm-base-plus_8s_1chunks.npz"
)

VAL_CACHE = CACHE_DIR / (
    "val_microsoft_wavlm-base-plus_8s_1chunks.npz"
)

CKPT_PATH = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.pt"
)

SCALER_PATH = CKPT_PATH.with_suffix(".scaler.joblib")

EPOCHS = 15
BATCH_SIZE = 256
LEARNING_RATE = 1e-3
HIDDEN_DIM = 256
DROPOUT = 0.2
SEED = 42


# =========================
# seed / device
# =========================

np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)


# =========================
# 캐시 확인
# =========================

assert TRAIN_CACHE.exists(), (
    f"Train cache가 없습니다: {TRAIN_CACHE}"
)

assert VAL_CACHE.exists(), (
    f"Validation cache가 없습니다: {VAL_CACHE}"
)

print("✅ Train cache 확인:", TRAIN_CACHE)
print("✅ Val cache 확인  :", VAL_CACHE)


# =========================
# embedding 로드
# =========================

print("\nEmbedding cache 로드 중...")

train_data = np.load(TRAIN_CACHE)
val_data = np.load(VAL_CACHE)

x_train = train_data["x"].astype(np.float32)
y_train = train_data["y"].astype(np.int64)

x_val = val_data["x"].astype(np.float32)
y_val = val_data["y"].astype(np.int64)

print("Train embedding shape:", x_train.shape)
print("Train label shape    :", y_train.shape)

print("Val embedding shape  :", x_val.shape)
print("Val label shape      :", y_val.shape)


# =========================
# 데이터 sanity check
# =========================

print("\nTrain label distribution:")
unique, counts = np.unique(y_train, return_counts=True)

for u, c in zip(unique, counts):
    print(f"class {u}: {c}")

print("\nValidation label distribution:")
unique, counts = np.unique(y_val, return_counts=True)

for u, c in zip(unique, counts):
    print(f"class {u}: {c}")


# =========================
# StandardScaler
# =========================

scaler = StandardScaler()

x_train = scaler.fit_transform(
    x_train
).astype(np.float32)

x_val = scaler.transform(
    x_val
).astype(np.float32)


# =========================
# Dataset / DataLoader
# =========================

train_dataset = TensorDataset(
    torch.from_numpy(x_train),
    torch.from_numpy(y_train),
)

val_dataset = TensorDataset(
    torch.from_numpy(x_val),
    torch.from_numpy(y_val),
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available(),
)


# =========================
# Classification Head
# =========================

class Head(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=256,
        dropout=0.2,
    ):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 2),
        )

    def forward(self, x):
        return self.net(x)


model = Head(
    input_dim=x_train.shape[1],
    hidden_dim=HIDDEN_DIM,
    dropout=DROPOUT,
).to(device)


# =========================
# 평가 함수
# =========================

def evaluate(model, loader):

    model.eval()

    all_pred = []
    all_true = []

    with torch.inference_mode():

        for x, y in loader:

            x = x.to(
                device,
                non_blocking=True
            )

            logits = model(x)

            pred = (
                logits
                .argmax(dim=1)
                .cpu()
                .numpy()
            )

            all_pred.extend(pred.tolist())
            all_true.extend(y.numpy().tolist())

    acc = accuracy_score(
        all_true,
        all_pred
    )

    macro_f1 = f1_score(
        all_true,
        all_pred,
        average="macro"
    )

    return (
        acc,
        macro_f1,
        np.asarray(all_true),
        np.asarray(all_pred),
    )


# =========================
# 학습 준비
# =========================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE
)

criterion = nn.CrossEntropyLoss()

best_f1 = -1.0
best_acc = -1.0
best_epoch = -1

CKPT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)


# =========================
# 학습
# =========================

print(
    "\n=== Frozen WavLM embedding "
    "+ classification head 학습 ==="
)

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for x, y in train_loader:

        x = x.to(
            device,
            non_blocking=True
        )

        y = y.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = model(x)

        loss = criterion(
            logits,
            y
        )

        loss.backward()
        optimizer.step()

        running_loss += (
            loss.item()
            * y.size(0)
        )

        correct += (
            logits.argmax(dim=1)
            == y
        ).sum().item()

        total += y.size(0)

    train_loss = (
        running_loss
        / total
    )

    train_acc = (
        correct
        / total
    )

    (
        val_acc,
        val_f1,
        _,
        _
    ) = evaluate(
        model,
        val_loader
    )

    print(
        f"epoch={epoch:02d} "
        f"train_loss={train_loss:.4f} "
        f"train_acc={train_acc:.4f} "
        f"val_acc={val_acc:.4f} "
        f"val_macro_f1={val_f1:.4f}"
    )

    # best macro F1 기준 저장
    if val_f1 > best_f1:

        best_f1 = val_f1
        best_acc = val_acc
        best_epoch = epoch

        torch.save(
            {
                "model_name":
                    "microsoft/wavlm-base-plus",

                "head_state_dict":
                    model.state_dict(),

                "input_dim":
                    x_train.shape[1],

                "hidden_dim":
                    HIDDEN_DIM,

                "dropout":
                    DROPOUT,

                "max_seconds":
                    8,

                "chunks_per_call":
                    1,

                "best_epoch":
                    best_epoch,

                "best_val_accuracy":
                    best_acc,

                "best_val_macro_f1":
                    best_f1,
            },
            CKPT_PATH,
        )

        joblib.dump(
            scaler,
            SCALER_PATH
        )

        print(
            "  ✅ Best checkpoint 저장:",
            CKPT_PATH
        )


# =========================
# BEST 모델 다시 로드
# =========================

checkpoint = torch.load(
    CKPT_PATH,
    map_location=device,
    weights_only=False,
)

model.load_state_dict(
    checkpoint["head_state_dict"]
)


# =========================
# 최종 Validation 평가
# =========================

(
    final_acc,
    final_f1,
    y_true,
    y_pred,
) = evaluate(
    model,
    val_loader
)

print("\n============================")
print(" BEST VALIDATION RESULT")
print("============================")

print(
    "Best epoch          :",
    checkpoint["best_epoch"]
)

print(
    f"Validation accuracy : "
    f"{final_acc:.4f}"
)

print(
    f"Validation macro F1 : "
    f"{final_f1:.4f}"
)

print("\nClassification Report")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=[
            "class 0",
            "class 1"
        ],
        digits=4,
    )
)

print("\n✅ checkpoint:")
print(CKPT_PATH)

print("\n✅ scaler:")
print(SCALER_PATH)
```

```python
%cd /content/drive/MyDrive/mission1

import joblib
import numpy as np
import torch

from pathlib import Path
from sklearn.metrics import accuracy_score, f1_score, classification_report

from train_mfcc_rf import collect_split_pairs, build_features
from train_wavlm import Head
from common import INDEX_TO_GENDER


# =========================================================
# 경로 설정
# =========================================================

DATA_ROOT = Path("/content/drive/MyDrive/data(inho)")

RF_CKPT = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_mfcc_rf.joblib"
)

WAVLM_VAL_CACHE = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache/"
    "val_microsoft_wavlm-base-plus_8s_1chunks.npz"
)

WAVLM_CKPT = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.pt"
)

WAVLM_SCALER = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.scaler.joblib"
)


# =========================================================
# 1. RandomForest 모델 로드
# =========================================================

rf_data = joblib.load(RF_CKPT)

rf_model = rf_data["classifier"]
n_mfcc = rf_data["n_mfcc"]
use_f0 = rf_data["use_f0"]
timestamp_unit = rf_data["timestamp_unit"]

print("=== RF 모델 설정 ===")
print("n_mfcc        :", n_mfcc)
print("use_f0        :", use_f0)
print("timestamp_unit:", timestamp_unit)
print("classes       :", rf_model.classes_)


# =========================================================
# 2. Validation MFCC + F0 feature 추출
# =========================================================

print("\n=== Validation MFCC+F0 feature 추출 ===")

val_pairs = collect_split_pairs(
    DATA_ROOT,
    "Validation"
)

x_val_rf, y_val_rf = build_features(
    val_pairs,
    n_mfcc=n_mfcc,
    use_f0=use_f0,
    timestamp_unit=timestamp_unit,
    desc="Validation RF feature",
)

print("\nRF feature shape:", x_val_rf.shape)
print("RF label shape  :", y_val_rf.shape)


# =========================================================
# 3. RF 확률 예측
# =========================================================

rf_prob = rf_model.predict_proba(x_val_rf)

print("RF probability shape:", rf_prob.shape)


# =========================================================
# 4. WavLM Validation embedding 로드
# =========================================================

wavlm_data = np.load(WAVLM_VAL_CACHE)

x_val_wavlm = wavlm_data["x"].astype(np.float32)
y_val_wavlm = wavlm_data["y"].astype(np.int64)

print("\nWavLM embedding shape:", x_val_wavlm.shape)
print("WavLM label shape    :", y_val_wavlm.shape)


# =========================================================
# 5. Validation 순서 일치 확인
# =========================================================

assert len(y_val_rf) == len(y_val_wavlm), (
    "RF와 WavLM Validation 샘플 수가 다릅니다."
)

assert np.array_equal(y_val_rf, y_val_wavlm), (
    "RF와 WavLM Validation 순서가 다릅니다."
)

print("\n✅ RF / WavLM Validation 순서 일치")


# =========================================================
# 6. WavLM scaler 적용
# =========================================================

scaler = joblib.load(WAVLM_SCALER)

x_val_scaled = scaler.transform(
    x_val_wavlm
).astype(np.float32)


# =========================================================
# 7. WavLM classification head 로드
# =========================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

checkpoint = torch.load(
    WAVLM_CKPT,
    map_location=device,
    weights_only=False,
)

wavlm_model = Head(
    input_dim=checkpoint["input_dim"],
    hidden_dim=checkpoint["hidden_dim"],
    dropout=checkpoint["dropout"],
).to(device)

wavlm_model.load_state_dict(
    checkpoint["head_state_dict"]
)

wavlm_model.eval()


# =========================================================
# 8. WavLM 확률 예측
# =========================================================

wavlm_probs = []

batch_size = 512

with torch.inference_mode():

    for start in range(
        0,
        len(x_val_scaled),
        batch_size
    ):

        x = torch.from_numpy(
            x_val_scaled[start:start + batch_size]
        ).to(device)

        logits = wavlm_model(x)

        prob = torch.softmax(
            logits,
            dim=1
        )

        wavlm_probs.append(
            prob.cpu().numpy()
        )

wavlm_prob = np.vstack(wavlm_probs)

print("WavLM probability shape:", wavlm_prob.shape)


# =========================================================
# 9. 단독 모델 성능 확인
# =========================================================

y_true = y_val_wavlm

rf_pred = rf_prob.argmax(axis=1)
wavlm_pred = wavlm_prob.argmax(axis=1)

rf_acc = accuracy_score(
    y_true,
    rf_pred
)

rf_f1 = f1_score(
    y_true,
    rf_pred,
    average="macro"
)

wavlm_acc = accuracy_score(
    y_true,
    wavlm_pred
)

wavlm_f1 = f1_score(
    y_true,
    wavlm_pred,
    average="macro"
)

print("\n============================")
print(" 단독 모델 성능")
print("============================")

print(
    f"RF     | acc={rf_acc:.4f} "
    f"| macro_f1={rf_f1:.4f}"
)

print(
    f"WavLM  | acc={wavlm_acc:.4f} "
    f"| macro_f1={wavlm_f1:.4f}"
)


# =========================================================
# 10. Late Fusion
# =========================================================

print("\n============================")
print(" Late Fusion")
print("============================")

results = []

for wavlm_weight in np.arange(
    0.0,
    1.01,
    0.05
):

    rf_weight = 1.0 - wavlm_weight

    fusion_prob = (
        wavlm_weight * wavlm_prob
        + rf_weight * rf_prob
    )

    fusion_pred = fusion_prob.argmax(
        axis=1
    )

    acc = accuracy_score(
        y_true,
        fusion_pred
    )

    macro_f1 = f1_score(
        y_true,
        fusion_pred,
        average="macro"
    )

    results.append(
        (
            wavlm_weight,
            rf_weight,
            acc,
            macro_f1,
        )
    )

    print(
        f"WavLM={wavlm_weight:.2f} "
        f"RF={rf_weight:.2f} "
        f"| acc={acc:.4f} "
        f"| macro_f1={macro_f1:.4f}"
    )


# =========================================================
# 11. 최고 Fusion 조합 찾기
# =========================================================

best = max(
    results,
    key=lambda x: x[3]
)

best_wavlm_weight = best[0]
best_rf_weight = best[1]
best_acc = best[2]
best_f1 = best[3]

print("\n============================")
print(" BEST FUSION")
print("============================")

print(
    f"WavLM weight : {best_wavlm_weight:.2f}"
)

print(
    f"RF weight    : {best_rf_weight:.2f}"
)

print(
    f"Accuracy     : {best_acc:.4f}"
)

print(
    f"Macro F1     : {best_f1:.4f}"
)


# =========================================================
# 12. 최고 Fusion classification report
# =========================================================

best_prob = (
    best_wavlm_weight * wavlm_prob
    + best_rf_weight * rf_prob
)

best_pred = best_prob.argmax(
    axis=1
)

target_names = [
    INDEX_TO_GENDER[i]
    for i in range(len(INDEX_TO_GENDER))
]

print("\n=== BEST FUSION Classification Report ===")

print(
    classification_report(
        y_true,
        best_pred,
        target_names=target_names,
        digits=4,
        zero_division=0,
    )
)
```

```python
%cd /content/drive/MyDrive/mission1

import os
import json
import torch
import numpy as np

from pathlib import Path
from transformers import AutoFeatureExtractor, AutoModel
from tqdm.auto import tqdm

from train_wavlm import collect_split_pairs
from common import SAMPLE_RATE, load_annotation, load_caller_audio


# =========================================================
# 설정
# =========================================================

MODEL_NAME = "microsoft/wavlm-base-plus"

DATA_ROOT = Path(
    "/content/drive/MyDrive/data(inho)"
)

SAVE_DIR = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache"
)

FINAL_CACHE = SAVE_DIR / (
    "val_microsoft_wavlm-base-plus_8s_3chunks.npz"
)

PROGRESS_FILE = SAVE_DIR / (
    "val_microsoft_wavlm-base-plus_8s_3chunks_progress.npz"
)

META_FILE = SAVE_DIR / (
    "val_microsoft_wavlm-base-plus_8s_3chunks_progress.json"
)

MAX_SECONDS = 8
CHUNKS_PER_CALL = 3

SAVE_EVERY = 200

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

SAVE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =========================================================
# 3-chunk 함수
# =========================================================

def make_three_chunks(
    waveform,
    max_samples,
):
    waveform = torch.as_tensor(
        waveform,
        dtype=torch.float32
    ).flatten()

    n = waveform.numel()

    # 8초 이하 → padding 후 1번만 처리
    if n <= max_samples:

        if n < max_samples:
            waveform = torch.nn.functional.pad(
                waveform,
                (0, max_samples - n)
            )

        return [waveform]

    # 8초 초과 → 앞 / 가운데 / 뒤 3개
    max_start = n - max_samples

    starts = [
        0,
        max_start // 2,
        max_start
    ]

    return [
        waveform[
            start:start + max_samples
        ]
        for start in starts
    ]


# =========================================================
# Validation pair 수집
# =========================================================

print("\nValidation 데이터 찾는 중...")

val_samples = collect_split_pairs(
    DATA_ROOT,
    "Validation"
)

print(
    f"Validation samples: "
    f"{len(val_samples):,}"
)


# =========================================================
# 이미 최종 cache 있으면 종료
# =========================================================

if FINAL_CACHE.exists():

    print("\n✅ 이미 최종 3-chunk cache가 존재합니다.")
    print(FINAL_CACHE)

else:

    # =====================================================
    # 이어서 실행할 progress 확인
    # =====================================================

    embeddings = []
    labels = []
    processed_stems = []

    start_index = 0

    if (
        PROGRESS_FILE.exists()
        and META_FILE.exists()
    ):

        print(
            "\n✅ 기존 progress 발견."
            " 이어서 실행합니다."
        )

        progress = np.load(
            PROGRESS_FILE
        )

        embeddings = list(
            progress["x"]
        )

        labels = list(
            progress["y"]
        )

        with open(
            META_FILE,
            "r",
            encoding="utf-8"
        ) as f:

            meta = json.load(f)

        processed_stems = meta[
            "processed_stems"
        ]

        start_index = len(
            processed_stems
        )

        print(
            f"이미 완료: "
            f"{start_index:,}"
        )

    else:

        print(
            "\n새 extraction을 시작합니다."
        )


    # =====================================================
    # WavLM 로드
    # =====================================================

    print("\nWavLM 로드 중...")

    feature_extractor = (
        AutoFeatureExtractor
        .from_pretrained(
            MODEL_NAME
        )
    )

    encoder = (
        AutoModel
        .from_pretrained(
            MODEL_NAME
        )
        .to(device)
    )

    encoder.eval()

    for p in encoder.parameters():
        p.requires_grad = False

    print("WavLM 로드 완료.")


    # =====================================================
    # extraction
    # =====================================================

    max_samples = int(
        SAMPLE_RATE
        * MAX_SECONDS
    )

    remaining = val_samples[
        start_index:
    ]

    progress_bar = tqdm(
        remaining,
        initial=start_index,
        total=len(val_samples),
        desc="Val WavLM 3-chunk"
    )

    for idx, sample in enumerate(
        progress_bar,
        start=start_index
    ):

        audio_path = sample[0]
        json_path = sample[1]

        annotation = load_annotation(
            json_path
        )

        gender = str(
            annotation.get(
                "gender",
                ""
            )
        ).strip().upper()

        # 기존 train_wavlm의 label 순서와 동일
        if gender == "M":
            label = 0
        elif gender == "F":
            label = 1
        else:
            raise ValueError(
                f"잘못된 gender: "
                f"{json_path} -> "
                f"{gender}"
            )


        # ---------------------------------------------
        # caller audio
        # ---------------------------------------------

        waveform = load_caller_audio(
            audio_path,
            annotation,
            timestamp_unit="auto"
        )


        # ---------------------------------------------
        # 앞 / 중간 / 뒤 8초
        # ---------------------------------------------

        chunks = make_three_chunks(
            waveform,
            max_samples
        )

        chunk_embeddings = []


        # ---------------------------------------------
        # WavLM 각각 통과
        # ---------------------------------------------

        for chunk in chunks:

            audio_np = (
                chunk
                .detach()
                .cpu()
                .numpy()
            )

            inputs = (
                feature_extractor(
                    audio_np,
                    sampling_rate=SAMPLE_RATE,
                    return_tensors="pt",
                    padding=True
                )
            )

            input_values = (
                inputs[
                    "input_values"
                ]
                .to(device)
            )

            attention_mask = (
                inputs.get(
                    "attention_mask"
                )
            )

            if attention_mask is not None:
                attention_mask = (
                    attention_mask
                    .to(device)
                )


            with torch.inference_mode():

                outputs = encoder(
                    input_values=
                        input_values,
                    attention_mask=
                        attention_mask
                )

                hidden = (
                    outputs
                    .last_hidden_state
                )


                # -------------------------------------
                # masked mean pooling
                # -------------------------------------

                if attention_mask is None:

                    pooled = (
                        hidden
                        .mean(dim=1)
                    )

                else:

                    feature_mask = (
                        encoder
                        ._get_feature_vector_attention_mask(
                            hidden.shape[1],
                            attention_mask
                        )
                    )

                    mask = (
                        feature_mask
                        .unsqueeze(-1)
                        .to(
                            hidden.dtype
                        )
                    )

                    pooled = (
                        (
                            hidden
                            * mask
                        )
                        .sum(dim=1)
                        /
                        mask
                        .sum(dim=1)
                        .clamp(
                            min=1.0
                        )
                    )


            chunk_embeddings.append(
                pooled
                .squeeze(0)
                .cpu()
                .numpy()
            )


        # ---------------------------------------------
        # 세 chunk embedding 평균
        # ---------------------------------------------

        call_embedding = np.mean(
            np.stack(
                chunk_embeddings,
                axis=0
            ),
            axis=0
        ).astype(
            np.float32
        )


        embeddings.append(
            call_embedding
        )

        labels.append(
            label
        )

        processed_stems.append(
            audio_path.stem
        )


        # ---------------------------------------------
        # 중간 저장
        # ---------------------------------------------

        completed = idx + 1

        if (
            completed % SAVE_EVERY == 0
            or completed == len(
                val_samples
            )
        ):

            x_tmp = np.stack(
                embeddings,
                axis=0
            )

            y_tmp = np.asarray(
                labels,
                dtype=np.int64
            )

            np.savez_compressed(
                PROGRESS_FILE,
                x=x_tmp,
                y=y_tmp
            )

            with open(
                META_FILE,
                "w",
                encoding="utf-8"
            ) as f:

                json.dump(
                    {
                        "processed_stems":
                            processed_stems
                    },
                    f,
                    ensure_ascii=False,
                    indent=2
                )

            print(
                f"\n💾 Progress 저장: "
                f"{completed:,}/"
                f"{len(val_samples):,}"
            )


    # =====================================================
    # 최종 저장
    # =====================================================

    x_val_3chunk = np.stack(
        embeddings,
        axis=0
    )

    y_val_3chunk = np.asarray(
        labels,
        dtype=np.int64
    )

    np.savez_compressed(
        FINAL_CACHE,
        x=x_val_3chunk,
        y=y_val_3chunk
    )

    print(
        "\n✅ 3-chunk Validation "
        "embedding 완료"
    )

    print(
        "shape:",
        x_val_3chunk.shape
    )

    print(
        "labels:",
        y_val_3chunk.shape
    )

    print(
        "저장 위치:",
        FINAL_CACHE
    )


    # progress 파일 정리
    if PROGRESS_FILE.exists():
        PROGRESS_FILE.unlink()

    if META_FILE.exists():
        META_FILE.unlink()

    print(
        "\n✅ progress 파일 정리 완료"
    )
```

```python
%cd /content/drive/MyDrive/mission1

import numpy as np
import joblib
import torch

from pathlib import Path
from sklearn.metrics import accuracy_score, f1_score, classification_report

from train_wavlm import Head
from common import INDEX_TO_GENDER


# =========================================================
# 경로
# =========================================================

CACHE_1 = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache/"
    "val_microsoft_wavlm-base-plus_8s_1chunks.npz"
)

CACHE_3 = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache/"
    "val_microsoft_wavlm-base-plus_8s_3chunks.npz"
)

CKPT = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.pt"
)

SCALER_PATH = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.scaler.joblib"
)


# =========================================================
# 데이터 로드
# =========================================================

data1 = np.load(CACHE_1)
data3 = np.load(CACHE_3)

x1 = data1["x"].astype(np.float32)
y1 = data1["y"].astype(np.int64)

x3 = data3["x"].astype(np.float32)
y3 = data3["y"].astype(np.int64)

print("1-chunk:", x1.shape)
print("3-chunk:", x3.shape)

assert np.array_equal(y1, y3)

print("✅ label 순서 일치")


# =========================================================
# scaler
# =========================================================

scaler = joblib.load(SCALER_PATH)

x1_scaled = scaler.transform(x1).astype(np.float32)
x3_scaled = scaler.transform(x3).astype(np.float32)


# =========================================================
# WavLM head 로드
# =========================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

checkpoint = torch.load(
    CKPT,
    map_location=device,
    weights_only=False
)

model = Head(
    input_dim=checkpoint["input_dim"],
    hidden_dim=checkpoint["hidden_dim"],
    dropout=checkpoint["dropout"]
).to(device)

model.load_state_dict(
    checkpoint["head_state_dict"]
)

model.eval()


# =========================================================
# probability 함수
# =========================================================

def predict_prob(x):

    probs = []

    batch_size = 512

    with torch.inference_mode():

        for start in range(
            0,
            len(x),
            batch_size
        ):

            batch = torch.from_numpy(
                x[start:start + batch_size]
            ).to(device)

            logits = model(batch)

            prob = torch.softmax(
                logits,
                dim=1
            )

            probs.append(
                prob.cpu().numpy()
            )

    return np.vstack(probs)


# =========================================================
# 예측
# =========================================================

prob1 = predict_prob(x1_scaled)
prob3 = predict_prob(x3_scaled)

pred1 = prob1.argmax(axis=1)
pred3 = prob3.argmax(axis=1)


# =========================================================
# 성능
# =========================================================

acc1 = accuracy_score(y1, pred1)
f1_1 = f1_score(
    y1,
    pred1,
    average="macro"
)

acc3 = accuracy_score(y3, pred3)
f1_3 = f1_score(
    y3,
    pred3,
    average="macro"
)


print("\n============================")
print(" WavLM 비교")
print("============================")

print(
    f"1-chunk | "
    f"acc={acc1:.4f} | "
    f"macro_f1={f1_1:.4f}"
)

print(
    f"3-chunk | "
    f"acc={acc3:.4f} | "
    f"macro_f1={f1_3:.4f}"
)

print("\nAccuracy 변화:")
print(
    f"{acc1:.4f} → {acc3:.4f} "
    f"({(acc3-acc1)*100:+.2f}%p)"
)


print("\n=== 3-chunk Classification Report ===")

target_names = [
    INDEX_TO_GENDER[i]
    for i in range(len(INDEX_TO_GENDER))
]

print(
    classification_report(
        y3,
        pred3,
        target_names=target_names,
        digits=4,
        zero_division=0
    )
)
```

```python
%cd /content/drive/MyDrive/mission1

import numpy as np
import joblib
import torch

from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
)

from train_mfcc_rf import collect_split_pairs, build_features
from train_wavlm import Head
from common import INDEX_TO_GENDER


# =========================================================
# 경로
# =========================================================

DATA_ROOT = Path(
    "/content/drive/MyDrive/data(inho)"
)

RF_CKPT = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_mfcc_rf.joblib"
)

WAVLM_CACHE_3 = Path(
    "/content/drive/MyDrive/mission1/wavlm_embedding_cache/"
    "val_microsoft_wavlm-base-plus_8s_3chunks.npz"
)

WAVLM_CKPT = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.pt"
)

WAVLM_SCALER = Path(
    "/content/drive/MyDrive/saved_models/mission1/"
    "mission1_wavlm_1chunk.scaler.joblib"
)


# =========================================================
# 1. RF 로드
# =========================================================

rf_data = joblib.load(RF_CKPT)

rf_model = rf_data["classifier"]

print("=== RF ===")
print("use_f0:", rf_data["use_f0"])


# =========================================================
# 2. Validation RF feature 추출
# =========================================================
# 이전 runtime이 끊겼으므로 RF feature를 다시 만들어야 함.
# WavLM embedding은 다시 추출하지 않음.
# =========================================================

print("\nValidation RF feature 추출...")

val_pairs = collect_split_pairs(
    DATA_ROOT,
    "Validation"
)

x_val_rf, y_val_rf = build_features(
    val_pairs,
    n_mfcc=rf_data["n_mfcc"],
    use_f0=rf_data["use_f0"],
    timestamp_unit=rf_data["timestamp_unit"],
    desc="Validation RF feature",
)

rf_prob = rf_model.predict_proba(
    x_val_rf
)


# =========================================================
# 3. 3-chunk WavLM embedding 로드
# =========================================================

data3 = np.load(WAVLM_CACHE_3)

x3 = data3["x"].astype(
    np.float32
)

y3 = data3["y"].astype(
    np.int64
)

assert np.array_equal(
    y_val_rf,
    y3
)

print(
    "\n✅ RF / WavLM sample 순서 일치"
)


# =========================================================
# 4. WavLM scaler
# =========================================================

scaler = joblib.load(
    WAVLM_SCALER
)

x3_scaled = scaler.transform(
    x3
).astype(np.float32)


# =========================================================
# 5. WavLM head
# =========================================================

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

checkpoint = torch.load(
    WAVLM_CKPT,
    map_location=device,
    weights_only=False
)

model = Head(
    input_dim=checkpoint["input_dim"],
    hidden_dim=checkpoint["hidden_dim"],
    dropout=checkpoint["dropout"]
).to(device)

model.load_state_dict(
    checkpoint["head_state_dict"]
)

model.eval()


# =========================================================
# 6. 3-chunk WavLM probability
# =========================================================

wavlm_probs = []

batch_size = 512

with torch.inference_mode():

    for start in range(
        0,
        len(x3_scaled),
        batch_size
    ):

        batch = torch.from_numpy(
            x3_scaled[
                start:start + batch_size
            ]
        ).to(device)

        logits = model(batch)

        prob = torch.softmax(
            logits,
            dim=1
        )

        wavlm_probs.append(
            prob.cpu().numpy()
        )

wavlm_prob = np.vstack(
    wavlm_probs
)


# =========================================================
# 7. 단독 성능 확인
# =========================================================

y_true = y3

rf_pred = rf_prob.argmax(axis=1)
wavlm_pred = wavlm_prob.argmax(axis=1)

rf_acc = accuracy_score(
    y_true,
    rf_pred
)

wavlm_acc = accuracy_score(
    y_true,
    wavlm_pred
)

print("\n============================")
print(" 단독 모델")
print("============================")

print(
    f"RF             : "
    f"{rf_acc:.4f}"
)

print(
    f"WavLM 3-chunk : "
    f"{wavlm_acc:.4f}"
)


# =========================================================
# 8. Fusion 탐색
# =========================================================
# WavLM weight를 0.00 ~ 1.00까지
# 0.01 간격으로 탐색
# =========================================================

print("\n============================")
print(" 3-chunk + RF Fusion")
print("============================")

results = []

for wavlm_weight in np.arange(
    0.00,
    1.001,
    0.01
):

    rf_weight = (
        1.0 - wavlm_weight
    )

    fusion_prob = (
        wavlm_weight * wavlm_prob
        +
        rf_weight * rf_prob
    )

    fusion_pred = (
        fusion_prob.argmax(
            axis=1
        )
    )

    acc = accuracy_score(
        y_true,
        fusion_pred
    )

    f1 = f1_score(
        y_true,
        fusion_pred,
        average="macro"
    )

    results.append(
        (
            wavlm_weight,
            rf_weight,
            acc,
            f1
        )
    )


# =========================================================
# 9. Accuracy 기준 정렬
# =========================================================

results_sorted = sorted(
    results,
    key=lambda x: (
        x[2],   # Accuracy
        x[3],   # 동률이면 Macro F1
    ),
    reverse=True
)


print("\n=== TOP 10 ===")

for result in results_sorted[:10]:

    w, r, acc, f1 = result

    print(
        f"WavLM={w:.2f} "
        f"RF={r:.2f} "
        f"| acc={acc:.4f} "
        f"| macro_f1={f1:.4f}"
    )


# =========================================================
# 10. BEST
# =========================================================

best = results_sorted[0]

best_wavlm = best[0]
best_rf = best[1]
best_acc = best[2]
best_f1 = best[3]

print("\n============================")
print(" 🏆 BEST FUSION")
print("============================")

print(
    f"WavLM weight : "
    f"{best_wavlm:.2f}"
)

print(
    f"RF weight    : "
    f"{best_rf:.2f}"
)

print(
    f"Accuracy     : "
    f"{best_acc:.4f}"
)

print(
    f"Macro F1     : "
    f"{best_f1:.4f}"
)


# =========================================================
# 11. Classification Report
# =========================================================

best_prob = (
    best_wavlm * wavlm_prob
    +
    best_rf * rf_prob
)

best_pred = best_prob.argmax(
    axis=1
)

target_names = [
    INDEX_TO_GENDER[i]
    for i in range(
        len(INDEX_TO_GENDER)
    )
]

print(
    "\n=== BEST Classification Report ==="
)

print(
    classification_report(
        y_true,
        best_pred,
        target_names=target_names,
        digits=4,
        zero_division=0
    )
)
```

```python
%%writefile /content/drive/MyDrive/mission1/mission1_submission/inference.py

import argparse
from pathlib import Path

import joblib
import numpy as np
import torch
from transformers import AutoFeatureExtractor, AutoModel

from common import (
    SAMPLE_RATE,
    INDEX_TO_GENDER,
    load_annotation,
    load_caller_audio,
)

from model import WavLMHead


MODEL_NAME = "microsoft/wavlm-base-plus"
MAX_SECONDS = 8


# =========================================================
# 앞 / 중간 / 뒤 3-chunk
# =========================================================

def make_three_chunks(waveform, max_samples):

    waveform = torch.as_tensor(
        waveform,
        dtype=torch.float32
    ).flatten()

    n = waveform.numel()

    # 8초 이하이면 한 번만 처리
    if n <= max_samples:

        if n < max_samples:
            waveform = torch.nn.functional.pad(
                waveform,
                (0, max_samples - n)
            )

        return [waveform]

    # 8초 초과
    max_start = n - max_samples

    starts = [
        0,
        max_start // 2,
        max_start,
    ]

    return [
        waveform[start:start + max_samples]
        for start in starts
    ]


# =========================================================
# WavLM embedding
# =========================================================

def extract_embedding(
    waveform,
    feature_extractor,
    encoder,
    device,
):

    max_samples = int(
        SAMPLE_RATE * MAX_SECONDS
    )

    chunks = make_three_chunks(
        waveform,
        max_samples
    )

    embeddings = []

    for chunk in chunks:

        audio_np = (
            chunk
            .detach()
            .cpu()
            .numpy()
        )

        inputs = feature_extractor(
            audio_np,
            sampling_rate=SAMPLE_RATE,
            return_tensors="pt",
            padding=True,
        )

        input_values = (
            inputs["input_values"]
            .to(device)
        )

        attention_mask = inputs.get(
            "attention_mask"
        )

        if attention_mask is not None:
            attention_mask = (
                attention_mask.to(device)
            )

        with torch.inference_mode():

            outputs = encoder(
                input_values=input_values,
                attention_mask=attention_mask,
            )

            hidden = outputs.last_hidden_state

            if attention_mask is None:

                pooled = hidden.mean(dim=1)

            else:

                feature_mask = (
                    encoder
                    ._get_feature_vector_attention_mask(
                        hidden.shape[1],
                        attention_mask,
                    )
                )

                mask = (
                    feature_mask
                    .unsqueeze(-1)
                    .to(hidden.dtype)
                )

                pooled = (
                    (hidden * mask).sum(dim=1)
                    /
                    mask.sum(dim=1).clamp(min=1.0)
                )

        embeddings.append(
            pooled
            .squeeze(0)
            .cpu()
            .numpy()
        )

    # 앞 / 중간 / 뒤 embedding 평균
    embedding = np.mean(
        np.stack(
            embeddings,
            axis=0
        ),
        axis=0,
    )

    return embedding.astype(
        np.float32
    )


# =========================================================
# Main
# =========================================================

def main():

    parser = argparse.ArgumentParser()

    parser.add_argument(
        "--audio_dir",
        required=True
    )

    parser.add_argument(
        "--label_dir",
        required=True
    )

    parser.add_argument(
        "--ckpt_path",
        required=True
    )

    parser.add_argument(
        "--output",
        required=True
    )

    args = parser.parse_args()


    audio_dir = Path(args.audio_dir)
    label_dir = Path(args.label_dir)
    ckpt_dir = Path(args.ckpt_path)
    output_path = Path(args.output)


    # -----------------------------------------------------
    # checkpoint
    # -----------------------------------------------------

    head_path = (
        ckpt_dir
        / "mission1_wavlm_1chunk.pt"
    )

    scaler_path = (
        ckpt_dir
        / "mission1_wavlm_1chunk.scaler.joblib"
    )


    if not head_path.exists():
        raise FileNotFoundError(
            f"checkpoint 없음: {head_path}"
        )

    if not scaler_path.exists():
        raise FileNotFoundError(
            f"scaler 없음: {scaler_path}"
        )


    # -----------------------------------------------------
    # Device
    # -----------------------------------------------------

    device = torch.device(
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    print("Device:", device)


    # -----------------------------------------------------
    # WavLM
    # -----------------------------------------------------

    print("WavLM 로드 중...")

    feature_extractor = (
        AutoFeatureExtractor
        .from_pretrained(
            MODEL_NAME
        )
    )

    encoder = (
        AutoModel
        .from_pretrained(
            MODEL_NAME
        )
        .to(device)
    )

    encoder.eval()

    print("WavLM 로드 완료.")


    # -----------------------------------------------------
    # scaler
    # -----------------------------------------------------

    scaler = joblib.load(
        scaler_path
    )


    # -----------------------------------------------------
    # head
    # -----------------------------------------------------

    checkpoint = torch.load(
        head_path,
        map_location=device,
        weights_only=False,
    )

    model = WavLMHead(
        input_dim=checkpoint["input_dim"],
        hidden_dim=checkpoint["hidden_dim"],
        dropout=checkpoint["dropout"],
    ).to(device)

    model.load_state_dict(
        checkpoint["head_state_dict"]
    )

    model.eval()


    # -----------------------------------------------------
    # WAV 목록
    # -----------------------------------------------------

    audio_files = sorted(
        audio_dir.rglob("*.wav")
    )

    print(
        f"Audio files: {len(audio_files):,}"
    )


    predictions = []


    # -----------------------------------------------------
    # inference
    # -----------------------------------------------------

    for i, audio_path in enumerate(
        audio_files,
        start=1
    ):

        # 동일 stem JSON 검색
        candidates = list(
            label_dir.rglob(
                audio_path.stem + ".json"
            )
        )

        if len(candidates) == 0:
            raise FileNotFoundError(
                f"JSON 없음: {audio_path.name}"
            )

        json_path = candidates[0]

        annotation = load_annotation(
            json_path
        )

        # caller 부분만 사용
        waveform = load_caller_audio(
            audio_path,
            annotation,
            timestamp_unit="auto",
        )

        embedding = extract_embedding(
            waveform,
            feature_extractor,
            encoder,
            device,
        )

        x = scaler.transform(
            embedding.reshape(1, -1)
        ).astype(np.float32)

        x_tensor = torch.from_numpy(
            x
        ).to(device)

        with torch.inference_mode():

            logits = model(
                x_tensor
            )

            pred_idx = int(
                logits.argmax(
                    dim=1
                ).item()
            )

        gender = INDEX_TO_GENDER[
            pred_idx
        ]

        predictions.append(
            (
                audio_path.name,
                gender
            )
        )


        if (
            i % 100 == 0
            or i == len(audio_files)
        ):
            print(
                f"[{i:,}/{len(audio_files):,}]"
            )


    # -----------------------------------------------------
    # CSV
    # -----------------------------------------------------

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    import csv

    with open(
        output_path,
        "w",
        newline="",
        encoding="utf-8",
    ) as f:

        writer = csv.writer(f)

        writer.writerow(
            [
                "audio file name",
                "gender",
            ]
        )

        writer.writerows(
            predictions
        )


    print("\n✅ inference 완료")
    print("Output:", output_path)
    print("Rows:", len(predictions))


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/drive/MyDrive/mission1_submission/inference.py

import argparse
import csv
from pathlib import Path

import numpy as np
import torch
from transformers import AutoFeatureExtractor, AutoModel

from common import (
    SAMPLE_RATE,
    INDEX_TO_GENDER,
    load_annotation,
    load_caller_audio,
)

from model import WavLMHead


# =========================================================
# 3-chunk
# =========================================================

def make_three_chunks(waveform, max_samples):

    waveform = torch.as_tensor(
        waveform,
        dtype=torch.float32
    ).flatten()

    n = waveform.numel()

    # 8초 이하 → 한 번만 처리
    if n <= max_samples:

        if n < max_samples:

            waveform = torch.nn.functional.pad(
                waveform,
                (0, max_samples - n)
            )

        return [waveform]

    # 8초 초과 → 앞 / 중간 / 뒤
    max_start = n - max_samples

    starts = [
        0,
        max_start // 2,
        max_start,
    ]

    return [
        waveform[
            start:start + max_samples
        ]
        for start in starts
    ]


# =========================================================
# WavLM embedding
# =========================================================

def extract_embedding(
    waveform,
    feature_extractor,
    encoder,
    device,
    max_seconds,
):

    max_samples = int(
        SAMPLE_RATE * max_seconds
    )

    chunks = make_three_chunks(
        waveform,
        max_samples
    )

    chunk_embeddings = []

    for chunk in chunks:

        audio_np = (
            chunk
            .detach()
            .cpu()
            .numpy()
        )

        inputs = feature_extractor(
            audio_np,
            sampling_rate=SAMPLE_RATE,
            return_tensors="pt",
            padding=True,
        )

        input_values = (
            inputs["input_values"]
            .to(device)
        )

        attention_mask = inputs.get(
            "attention_mask"
        )

        if attention_mask is not None:

            attention_mask = (
                attention_mask
                .to(device)
            )

        with torch.inference_mode():

            outputs = encoder(
                input_values=input_values,
                attention_mask=attention_mask,
            )

            hidden = outputs.last_hidden_state

            # Validation 3-chunk embedding을 만들 때와
            # 동일한 pooling
            if attention_mask is None:

                pooled = hidden.mean(
                    dim=1
                )

            else:

                feature_mask = (
                    encoder
                    ._get_feature_vector_attention_mask(
                        hidden.shape[1],
                        attention_mask,
                    )
                )

                mask = (
                    feature_mask
                    .unsqueeze(-1)
                    .to(hidden.dtype)
                )

                pooled = (
                    (hidden * mask).sum(dim=1)
                    /
                    mask.sum(dim=1).clamp(
                        min=1.0
                    )
                )

        chunk_embeddings.append(
            pooled
            .squeeze(0)
            .cpu()
            .numpy()
        )

    # 앞 / 중간 / 뒤 embedding 평균
    embedding = np.mean(
        np.stack(
            chunk_embeddings,
            axis=0
        ),
        axis=0,
    )

    return embedding.astype(
        np.float32
    )


# =========================================================
# scaler
# =========================================================

def apply_scaler(
    embedding,
    scaler_mean,
    scaler_scale,
):

    # sklearn StandardScaler.transform과 동일:
    # (x - mean) / scale
    scaled = (
        embedding - scaler_mean
    ) / scaler_scale

    return scaled.astype(
        np.float32
    )


# =========================================================
# JSON 찾기
# =========================================================

def build_json_map(label_dir):

    json_files = sorted(
        label_dir.rglob("*.json")
    )

    json_map = {}

    for path in json_files:

        # 같은 stem이 중복되면 위험하므로 검사
        if path.stem in json_map:

            raise RuntimeError(
                f"중복 JSON stem 발견: "
                f"{path.stem}"
            )

        json_map[path.stem] = path

    return json_map


# =========================================================
# main
# =========================================================

def main():

    parser = argparse.ArgumentParser()

    parser.add_argument(
        "--audio_dir",
        required=True,
        type=str,
    )

    parser.add_argument(
        "--label_dir",
        required=True,
        type=str,
    )

    parser.add_argument(
        "--ckpt_path",
        required=True,
        type=str,
    )

    parser.add_argument(
        "--output",
        required=True,
        type=str,
    )

    args = parser.parse_args()


    # -----------------------------------------------------
    # Path
    # -----------------------------------------------------

    audio_dir = Path(
        args.audio_dir
    )

    label_dir = Path(
        args.label_dir
    )

    ckpt_path = Path(
        args.ckpt_path
    )

    output_path = Path(
        args.output
    )


    if not audio_dir.exists():

        raise FileNotFoundError(
            f"audio_dir 없음: "
            f"{audio_dir}"
        )

    if not label_dir.exists():

        raise FileNotFoundError(
            f"label_dir 없음: "
            f"{label_dir}"
        )

    if not ckpt_path.is_file():

        raise FileNotFoundError(
            f"checkpoint 파일 없음: "
            f"{ckpt_path}"
        )


    # -----------------------------------------------------
    # Device
    # -----------------------------------------------------

    device = torch.device(
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    print(
        f"Device: {device}"
    )


    # -----------------------------------------------------
    # checkpoint
    # -----------------------------------------------------

    print(
        "Checkpoint 로드 중..."
    )

    checkpoint = torch.load(
        ckpt_path,
        map_location="cpu",
        weights_only=False,
    )

    required_keys = [
        "head_state_dict",
        "input_dim",
        "hidden_dim",
        "dropout",
        "scaler_mean",
        "scaler_scale",
        "model_name",
        "max_seconds",
    ]

    for key in required_keys:

        if key not in checkpoint:

            raise KeyError(
                f"checkpoint에 "
                f"'{key}' 없음"
            )


    scaler_mean = np.asarray(
        checkpoint["scaler_mean"],
        dtype=np.float32,
    )

    scaler_scale = np.asarray(
        checkpoint["scaler_scale"],
        dtype=np.float32,
    )

    model_name = checkpoint[
        "model_name"
    ]

    max_seconds = float(
        checkpoint["max_seconds"]
    )


    print(
        "Model:",
        model_name
    )

    print(
        "Chunk strategy:",
        checkpoint.get(
            "inference_strategy",
            "front_middle_end_mean"
        )
    )


    # -----------------------------------------------------
    # WavLM
    # -----------------------------------------------------

    print(
        "\nWavLM encoder 로드 중..."
    )

    feature_extractor = (
        AutoFeatureExtractor
        .from_pretrained(
            model_name
        )
    )

    encoder = (
        AutoModel
        .from_pretrained(
            model_name
        )
        .to(device)
    )

    encoder.eval()

    for parameter in (
        encoder.parameters()
    ):
        parameter.requires_grad = False


    print(
        "WavLM encoder 로드 완료."
    )


    # -----------------------------------------------------
    # Head
    # -----------------------------------------------------

    head = WavLMHead(
        input_dim=int(
            checkpoint["input_dim"]
        ),
        hidden_dim=int(
            checkpoint["hidden_dim"]
        ),
        dropout=float(
            checkpoint["dropout"]
        ),
    ).to(device)

    head.load_state_dict(
        checkpoint[
            "head_state_dict"
        ]
    )

    head.eval()


    # -----------------------------------------------------
    # 파일 목록
    # -----------------------------------------------------

    audio_files = sorted(
        audio_dir.rglob("*.wav")
    )

    if len(audio_files) == 0:

        raise RuntimeError(
            f"WAV 파일이 없습니다: "
            f"{audio_dir}"
        )


    print(
        f"\nAudio files: "
        f"{len(audio_files):,}"
    )


    print(
        "JSON index 생성 중..."
    )

    json_map = build_json_map(
        label_dir
    )

    print(
        f"JSON files: "
        f"{len(json_map):,}"
    )


    # -----------------------------------------------------
    # inference
    # -----------------------------------------------------

    predictions = []

    for i, audio_path in enumerate(
        audio_files,
        start=1
    ):

        json_path = json_map.get(
            audio_path.stem
        )

        if json_path is None:

            raise FileNotFoundError(
                f"JSON 없음: "
                f"{audio_path.name}"
            )


        # ---------------------------------------------
        # annotation
        # ---------------------------------------------

        annotation = load_annotation(
            json_path
        )


        # ---------------------------------------------
        # 신고자 음성만 추출
        # ---------------------------------------------

        waveform = load_caller_audio(
            audio_path,
            annotation,
            timestamp_unit="auto",
        )


        # ---------------------------------------------
        # WavLM 3-chunk embedding
        # ---------------------------------------------

        embedding = extract_embedding(
            waveform,
            feature_extractor,
            encoder,
            device,
            max_seconds,
        )


        # ---------------------------------------------
        # scaler
        # ---------------------------------------------

        embedding_scaled = (
            apply_scaler(
                embedding,
                scaler_mean,
                scaler_scale,
            )
        )

        x = torch.from_numpy(
            embedding_scaled
        ).unsqueeze(0).to(
            device
        )


        # ---------------------------------------------
        # Head
        # ---------------------------------------------

        with torch.inference_mode():

            logits = head(x)

            pred_idx = int(
                logits.argmax(
                    dim=1
                ).item()
            )


        gender = INDEX_TO_GENDER[
            pred_idx
        ]


        predictions.append(
            [
                audio_path.name,
                gender,
            ]
        )


        if (
            i % 100 == 0
            or i == len(audio_files)
        ):

            print(
                f"[{i:,}/"
                f"{len(audio_files):,}]"
            )


    # -----------------------------------------------------
    # CSV
    # -----------------------------------------------------

    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    with open(
        output_path,
        "w",
        newline="",
        encoding="utf-8",
    ) as f:

        writer = csv.writer(f)

        writer.writerow(
            [
                "audio file name",
                "gender",
            ]
        )

        writer.writerows(
            predictions
        )


    print(
        "\n============================"
    )

    print(
        "✅ Mission 1 inference 완료"
    )

    print(
        "Output:",
        output_path
    )

    print(
        "Rows:",
        len(predictions)
    )

    print(
        "============================"
    )


if __name__ == "__main__":
    main()
```

```python
import csv
import json
from pathlib import Path

CSV_PATH = Path(
    "/content/mission1_smoke/mission1.csv"
)

LABEL_DIR = Path(
    "/content/mission1_smoke/label"
)

# CSV 읽기
with open(
    CSV_PATH,
    "r",
    encoding="utf-8"
) as f:
    rows = list(csv.DictReader(f))


print("=== CSV 검사 ===")
print("행 개수:", len(rows))
print("컬럼:", list(rows[0].keys()))

assert len(rows) == 5
assert list(rows[0].keys()) == [
    "audio file name",
    "gender"
]

print("✅ CSV 형식 정상")


# 실제 정답과 비교
correct = 0

print("\n=== Smoke Test 예측 ===")

for row in rows:

    audio_name = row["audio file name"]
    pred = row["gender"]

    json_path = (
        LABEL_DIR
        / Path(audio_name).with_suffix(".json").name
    )

    with open(
        json_path,
        "r",
        encoding="utf-8"
    ) as f:
        annotation = json.load(f)

    true_gender = str(
        annotation["gender"]
    ).strip().upper()

    is_correct = (
        pred == true_gender
    )

    correct += int(is_correct)

    print(
        f"{audio_name}\n"
        f"  정답={true_gender} | "
        f"예측={pred} | "
        f"{'✅' if is_correct else '❌'}"
    )


print("\n============================")
print(
    f"Smoke accuracy: "
    f"{correct}/{len(rows)} "
    f"({correct/len(rows):.1%})"
)
print("============================")
```
